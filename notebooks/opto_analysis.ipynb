```python
from concurrent.futures import ProcessPoolExecutor
from pathlib import Path

from sniffing import process_files
```

```python
# data_dir = Path("/mnt/r2d2/5_Projects/2_Optogenetics/PENTYL VS HEXYL discrimination/Raw data")
data_dir = Path("R:/5_Projects/2_Optogenetics/PENTYL VS HEXYL discrimination/Raw data")
experiment_dirs = list(data_dir.iterdir())
h5_files = {}

for _dir in experiment_dirs:
    if "z_" in _dir.stem.lower():
        continue
    h5_files[_dir.name] = list(_dir.glob("*.h5"))

experiments = list(h5_files.keys())

output_dir = data_dir.parent.joinpath("output")
output_dir.mkdir(exist_ok=True, parents=True)

ppe = ProcessPoolExecutor(max_workers=12)
```

```python
for exp in experiments:
    exp_output_dir = output_dir.joinpath(exp)
    exp_output_dir.mkdir(exist_ok=True, parents=True)
    ppe.submit(process_files.process_files, h5_files[exp], exp_output_dir, plot_figs=True)

ppe.shutdown()
```
